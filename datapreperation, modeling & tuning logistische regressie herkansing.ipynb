<span style = 'font-size:24px;'><b> Preperation, modeling & tuning</b></span>

```python
# Numerieke kolommen en categorische kolommen en het referentie niveau dat bij de categorische kollomen gedropt wordt.
numcols = ['Call Failure', 'Subscription Length', 'Seconds of Use', 'Frequency of use', 'Frequency of SMS']
faccols = ['Age Group', 'Complaints']
droplevels = [1, 0]
```

We maken 3 lijsten die het volgende doen:
- numcols bestaat uit alle numerieke kollomen in de trainset.
- faccols bestaat uit alle categorische kollomen in de trainset.
- droplevels bestaat uit waardes die in faccols gedropt moeten worden die het referentie niveau voorstellen.

```python
# Columntransformer voor de datapreperatie fase.
dataprep = ColumnTransformer([
    ('numeriek', StandardScaler(), numcols),
    ('categorisch', OneHotEncoder(drop = droplevels, sparse_output=False), faccols),
], remainder = 'drop', verbose_feature_names_out = False)

dataprep.set_output(transform = 'pandas')
dataprep
```

We voegen aan de columntransformer het volgende toe:
- Een standardscaler die de numerieke kollomen transformeert naar gestandadiseerde waardes.
- Een onehotencoder die de categorische kollomen transformeert waarbij het referentie niveau gedropt wordt.

```python
# We passen hier een pipeline toe voor de traindata en we voegen een logistische regressie toe.
pipeline = Pipeline([
    ('preprocessing', dataprep),
    ('model', LogisticRegression())
])

pipeline
```

- We voegen de data en het logistische regressiemodel toe aan de pipeline, zodat wij het straks kunnen fitten.

```python
# We definieren X_train en y_train en fitten het model op de data.

X_train = data_train.drop(columns = ['Churn'])
y_train = data_train['Churn']

model = pipeline.fit(X_train,y_train)

# We berekenen de accuracy  van het model.
y_pred = model.predict(X_train)
print(f'accuracy van het model:{accuracy_score(y_train,y_pred):.2f}')
```

- We definieren X_train als alle kollomen uit de dataset behalve de kolom Churn die samen de verklarende variabelen van het model zijn.
- y_train bestaat alleen uit de kolom Churn die onze responsvariabele is.
- Vervolgens wordt het model gefit op de data en de accuracy van het model wordt bepaald.

```python
# We tunen het model door de regularisatie term C te tunen.
# We beoordelen het model op basis van de accuracy.


# we zorgen voor een gevarieerde grid. 
params = {
 'model__C': [0.001,0.01,0.1,1,10,100]}

# hieronder wordt de gridsearch uitgevoerd en daarna gefit.
gridsearch = GridSearchCV(
    model,
    param_grid = params,
    scoring='accuracy'
    )
gridsearch.fit(X_train,y_train)

# we bepalen de accuracy en de optimale c waarde.
print(f'best accuracy:{gridsearch.best_score_:.3f}')
gridsearch.best_params_
```

- De eerste gridsearch die wij gebruiken is om de orde van grootte van de hyperparameter C te bepalen. We nemen machten van 10 startend vanaf 0.001 t/m 100.
- De uitkomst van deze gridsearch is 10 met een accuracy van 0.897 afgerond op 3 decimalen.
- Nu weten wij niet of er in het interval (1,10) en (10,100) een betere waarde van C te vinden is.
- In de onderstaande gridsearch gaan wij in bovenstaande interval verder zoeken.


```python
# de tweede gridsearch waarbij wij zoeken tussen 1 tot 10 en 10 tot 100.

# We stellen het grid samen.
params = {
 'model__C': np.linspace(1,100,100)}

# hieronder wordt de gridsearch uitgevoerd en daarna gefit.
gridsearch = GridSearchCV(
    model,
    param_grid = params,
    scoring='accuracy'
    )
gridsearch.fit(X_train,y_train)

# we bepalen de accuracy en de optimale C waarde uit de nieuwe gridsearch.
print(f'best accuracy:{gridsearch.best_score_:.3f}')
gridsearch.best_params_
```

- In de tweede gridsearch hierboven is te zien dat de optimale waarde van de hyperparameter C gelijk is aan 6.
- De accuracy van het model is met 0.001 verbeterd ten opzichte van de eerste gridsearch.
- De reden dat wij alleen kijken naar gehele C waarden in onze gridsearch is, omdat er geen verschil te zien is in accuracy tussen gehele C waarden en decimale C waarden.
- In totaal hebben wij dus 2 gridsearches gebruikt om de optimale C te bepalen.

```python
# In deze  cel worden 3  plots gemaakt de verdeling voorspelde kansen, een metric plot en een roc curve.
y_pred = gridsearch.predict_proba(X_train)

# Functie die een classificatie report omzet in rijen van een dataframe.
def make_row(cr):
    return {
        'Accuracy':  cr['accuracy'],
        'Recall':    cr['1']['recall'],
        'Precision': cr['1']['precision'],
        'F1 Score':  cr['1']['f1-score']
    }

thresholds = np.arange(0, 1, 0.01)
plot_data = pd.DataFrame([
    make_row(classification_report(
        y_train, 
        y_pred[:,1] > t, 
        zero_division=True, 
        output_dict=True)) for t in thresholds])
plot_data.index = thresholds

plt.figure(figsize=(12,4), layout='tight')

# Plot 1: Verdeling van de voorspelde kansen.
plt.subplot(1,3,1)
sns.kdeplot(data_train, x = y_pred[:,1], hue = 'Churn',fill = True)
plt.xlim(0,1)
plt.title('Verdeling voorspelde kansen')
plt.xlabel('Waarde van de kansen')
plt.ylabel('Dichtheid')

# Plot 2: Een metric plot.
nul_acc = 0
plt.subplot(1,3,2)
sns.lineplot(plot_data)
plt.plot([0,1], [nul_acc, nul_acc], 'k:', alpha = 0.5, label = 'Nul Acc.')
plt.xlabel('Threshold')
plt.legend()
plt.title('Metric Plot')
plt.ylabel('Waarde van de metric')


# Plot 3: Een ROC curve.
ax = plt.subplot(1,3,3)
roc = RocCurveDisplay.from_predictions(
    y_train, 
    y_pred=y_pred[:,1], 
    name = '', 
    plot_chance_level = True,
    ax=ax)
plt.fill_between(roc.fpr, roc.tpr, color = '0.8')
plt.title('ROC curve')
plt.xlabel('False Positive Rate (blijven)')
plt.ylabel('True Positive Rate (blijven)')
plt.show()
print(f'accuracy model: {accuracy_score(y_train,gridsearch.predict(X_train)):.2f}')
```

In de eerste plot is te zien dat de kans dat een klant blijft een hogere dichtheid heeft vergeleken met als ze opzeggen van 0 tot 0,45. Vanaf 0,45 tot 1 heeft hebben de kansen dat klanten opzeggen weer een hogere dichtheid.

In de tweede plot is te zien hoe de waarde van de metrics afhangt van de threshold. 
- Bij toenemende threshold wordt de recall kleiner en precision groter.
- Bij afnemenende threshild wordt de recall groter en precision kleiner.
- De F1-score zit tussen de recall en precision sinds het een gemiddelde is van de 2.

In de derde plot is een roc curve te zien van het model met een AUC (area under curve) die gelijk is aan 0.93.

```python
# In deze cel wordt een confusionmatrix gemaakt voor de variabele churn.
# Ook wordt een dataframe gemaakt die de waardes van de 
# accuracy, recall, precision en F1-score laat zien voor het logistische regressie model.

# Confusionmatrix maken.
y_pred = gridsearch.predict(X_train)
ConfusionMatrixDisplay.from_predictions(y_train,y_pred, cmap = 'Blues',display_labels = ['Blijven','Opzeggen'])
plt.title('Confusionmatrix van churn op de train data')
plt.xlabel('Voorspelde label')
plt.ylabel('Werkelijke label')
plt.show()

# Hier worden de accuracy, f1, precision en recall scores berekend en afgerond op 2 decimalen.
acc = round(accuracy_score(y_train, y_pred ), 2)
rec = round(recall_score(y_train, y_pred ), 2)
pre = round(precision_score(y_train, y_pred ), 2)
fsco = round(f1_score(y_train, y_pred ), 2)


# we maken een dictionary, zodat we er vervolgens een dataframe van kunnen maken.
table = {"Name": ["Accuracy", "Recall", "Precision", "F1-score"],
        "Value": [acc, rec, pre, fsco]}
df = pd.DataFrame(table)
print(df)
```

In de confusion matrix is te zien dat:
- 1667 klanten die blijven worden niet gebeld.
- 126 klanten die opzeggen worden gebeld (krijgen terecht het goedkope tarief).
- 190 klanten die opzeggen worden niet gebeld. 
- 12 klanten die blijven worden wel gebeld (krijgen onterecht het goedkopere tarief).

In de dataframe zijn de waarden per prestatie maat te zien:
- De accuracy van het model is 0,9.
- De recall van het model is 0,4.
- De precision van het model is 0,41.
- De F1-score van het model is 0,56.

Het model voorspelt over het algemeen goed of klanten willen blijven of willen opzeggen die te zien is aan de  hoge accuracy. Opvallend is dat het model zo kan voorspellen dat er weinig mensen onterecht het goedkope tarief krijgen te zien aan de hoge precision. Ook worden er veel klanten niet gebeld die wel willen opzeggen, waardoor deze klanten verloren gaan te zien aan de lage recall.
