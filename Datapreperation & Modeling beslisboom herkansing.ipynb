```python
# Numerieke kolommen en categorische kolommen en het referentie niveau dat gedropt wordt.
numcols = ['Call Failure','Subscription Length','Seconds of Use','Frequency of use','Frequency of SMS']
faccols = ['Age Group','Complaints']
droplevels = [1,0]
```

We maken 3 lijsten die het volgende doen:
- numcols bestaat uit alle numerieke kollomen in de trainset.
- faccols bestaat uit alle categorische kollomen in de trainset.
- droplevels bestaat uit waardes die in faccols gedropt moeten worden die het referentie niveau voorstellen.

```python

# Columntransformer voor de datapreperatie fase.
dataprep = ColumnTransformer([
    ('numeriek', StandardScaler(), numcols),
    ('categorisch', OneHotEncoder(drop = droplevels, sparse_output=False), faccols),
], remainder = 'drop', verbose_feature_names_out = False)

dataprep.set_output(transform='pandas')
dataprep
```

We voegen aan de columntransformer het volgende toe:
- Een standardscaler die de numerieke kollomen transformeert naar gestandadiseerde waardes.
- Een onehotencoder die de categorische kollomen transformeert waarbij het referentie niveau gedropt wordt.

```python

pipeline = Pipeline([('preprocessing', dataprep),
                  ('model', DecisionTreeClassifier(max_depth = 4,
                                                   min_samples_leaf=1,random_state = 1987))])
pipeline
```

Om het model te kunnen fitten wordt de data samen met de beslisboom toegevoegd aan de pipeline met de volgende hyperparameters: 
- max_depth = 4 (de boom heeft een maximale diepte van 4)
- min_samples_leaf = 1 (elke boom heeft minstens 1 sample) 
- random_state = 1987 voor reproduceerbaarheid van de resultaten. 

We kiezen een max_depth van 4, omdat het model complex genoeg is dat het kan leren va de data,maar niet zo complex dat het gaat overfitten.
We kiezen min_samples_leaf van 1 , zodat de boom altijd kan vertakken. Dit kan ervoor zorgen dat het model beter kan presteren op de metrics.

```python
X_train = data_train.drop(columns = ['Churn'])
y_train = data_train['Churn']

# de hyperparameters zijn zo gekozen,omdat het aanbevolen wordt door de userguide van sklearn.
model = pipeline.fit(X_train, y_train)
y_pred =model.predict(X_train)
print(f'accuracy model: {accuracy_score(y_train,y_pred):.2f}')
```

- We definieren X_train als alle kollomen uit de dataset behalve de kolom Churn die samen de verklarende variabelen van het model zijn.
- y_train bestaat alleen uit de kolom Churn die onze responsvariabele is.
- Vervolgens wordt het model gefit op de data en de accuracy van het model wordt bepaald.

```python
# In deze cel wordt een plot gemaakt van de boom.
# De diepte van de plot is 1 zodat de informatie per leaf-node duidelijk te zien is.

plot_tree(model.named_steps['model'],
            feature_names = data_train.columns,
            filled = True,
            proportion=True,
            label = 'root',max_depth = 1)
plt.title('Plot van de boom')
plt.show()
```

In de visualisatie is een deel van onze beslisboom te zien die de volgende informatie bevat per leaf:
- De voorwaarde waaronder gesplitst wordt.
- De gini-index van het model die de kwaliteit van een split bepaald.
- Het percentage van de data dat in die node zit.
- Value is een lijst met twee elementen waarbij het eerste elemen het percentage dat Churn = 0 en de tweede is het percentage dat Churn = 1. 

```python
# We halen alle unieke alphas uit de boom.

alphas = DecisionTreeClassifier(max_depth = 4,min_samples_leaf=1,random_state = 1987).fit(X_train, y_train).cost_complexity_pruning_path(X_train, y_train)
alphas = np.unique(alphas['ccp_alphas'])
alphas
```

We halen de alphas uit de boom om ze vervolgens te gebruiken voor de gridsearch hieronder.

```python
# Hieronder wordt een gridsearch uitgevoerd om de beste waarde van alpha te bepalen.
# Dit doen wij aan de hand van de accuracy.


# we tunen op basis van de alphas die uit de boom zijn gehaald en fitten de gridsearch op de data.
gridsearch = GridSearchCV(
    model,
    param_grid = {'model__ccp_alpha' : alphas
                  },
    scoring = 'accuracy',
    n_jobs=-2)

gridsearch.fit(X_train, y_train)

# we slaan de beste alpha en accuracy op en printen het uit.
best_alpha = gridsearch.best_params_['model__ccp_alpha']
best_score = gridsearch.best_score_
 
# we bepalen de accuracy en de optimale alpha waarde.
print(f'Best gevonden alpha: {best_alpha:.4e}, accuracy = {best_score:.2f}')
gridsearch.best_estimator_
```

We zien dat de optimale alpha van het model gelijk is aan 3.7802-e3 wat een kleine waarde is voor alpha. Ook te is te zien dat de accuracy van het model gelijk is aan 0,90.

```python
# In deze cel wordt een plot gemaakt waarbij de alpha uitgezet wordt tegen de accuracy.

sns.lineplot(gridsearch.cv_results_, x = 'param_model__ccp_alpha', y = 'mean_test_score', label = 'Gemid.')
for i in range(5):
    sns.lineplot(gridsearch.cv_results_, 
                 x = 'param_model__ccp_alpha', 
                 y = f'split{i}_test_score', 
                 alpha = 0.2, 
                 label = f'Fold {i+1}')
plt.plot(best_alpha, best_score, 'ro')
plt.xlim(0, 0.003)
plt.title('Accuracy uitgezet tegen cost complexity $\\alpha$')
plt.xlabel('Cost complexity $\\alpha$')
plt.ylabel('accuracy')
plt.show()
```

- In de visualisatie is te zien dat de accuracy niet/nauwelijks veranderd en ook denkt dat het geen alpha nodig heeft. in de plot is te zien dat de accuracy bijna niet veranderd, omdat het model denkt dat er geen alpha nodig is voor de beslisboom.

```python
# In deze  cel worden 2  plots gemaakt namelijk de verdeling van  voorspelde kansen en de roc-curve.
y_pred = gridsearch.predict_proba(X_train)

# Functie die een classificatie report omzet in rijen van een dataframe.
def make_row(cr):
    return {
        'Accuracy':  cr['accuracy'],
        'Recall':    cr['1']['recall'],
        'Precision': cr['1']['precision'],
        'F1 Score':  cr['1']['f1-score']
    }

thresholds = np.arange(0, 1, 0.01)
plot_data = pd.DataFrame([
    make_row(classification_report(
        y_train, 
        y_pred[:,1] > t, 
        zero_division=True, 
        output_dict=True)) for t in thresholds])
plot_data.index = thresholds

plt.figure(figsize=(12,4), layout='tight')

# Plot 1: Verdeling van de voorspelde kansen.
plt.subplot(1,3,1)
sns.kdeplot(data_train, x = y_pred[:,1], hue = 'Churn',fill = True)
plt.xlim(0,1)
plt.title('Verdeling voorspelde kansen')
plt.xlabel('Waarde van de kansen')
plt.ylabel('Dichtheid')
plt.legend(title = 'Churn',labels =['Blijven','Opzeggen'])


# Plot 2: Een ROC curve.
ax = plt.subplot(1,3,2)
roc = RocCurveDisplay.from_predictions(
    y_train, 
    y_pred=y_pred[:,1], 
    name = '', 
    plot_chance_level = True,
    ax=ax)
plt.fill_between(roc.fpr, roc.tpr, color = '0.8')
plt.title('ROC curve')
plt.xlabel('False Positive Rate (blijven)')
plt.ylabel('True Positive Rate (blijven)')
plt.show()
print(f'accuracy model: {accuracy_score(y_train,gridsearch.predict(X_train)):.2f}')

```

In de eerste plot is te zien dat de kans dat een klant blijft een hogere dichtheid heeft vergeleken met als ze opzeggen die te zien is aan de hoge pieken in de visualisatie.

In de tweede plot is een roc curve te zien van het model met een AUC (area under curve) die gelijk is aan 0,84.

```python
# In deze cel wordt een confusionmatrix gemaakt voor de variabele churn.
# Ook wordt een dataframe gemaakt die de waardes van de 
# accuracy, recall, precision en F1-score laat zien voor het logistische regressie model.


# Confusionmatrix maken.
y_pred = gridsearch.predict(X_train)
ConfusionMatrixDisplay.from_predictions(y_train,y_pred, cmap = 'Blues',display_labels = ['Blijven','Opzeggen'])
plt.title('Confusionmatrix van churn op de train data')
plt.xlabel('Voorspelde label')
plt.ylabel('Werkelijke label')
plt.show()

# Hier worden de accuracy, f1, precision en recall scores berekend en gegeven.
acc = round(accuracy_score(y_train, y_pred ), 2)
rec = round(recall_score(y_train, y_pred ), 2)
pre = round(precision_score(y_train, y_pred ), 2)
fsco = round(f1_score(y_train, y_pred ), 2)

# we maken een dictionary, zodat we er vervolgens een dataframe van kunnen maken.
table = {"Name": ["Accuracy", "Recall", "Precision", "F1-score"],
        "Value": [acc, rec, pre, fsco]}
df = pd.DataFrame(table)
print(df)
```

In de confusion matrix is te zien dat er over het algemeen goed voorspelt wordt dat:
- 1588 klanten die blijven worden niet gebeld.
- 222 klanten die opzeggen worden gebeld (krijgen terecht het goedkope tarief).
- 94 klanten die opzeggen  worden niet gebeld.
- 91 klanten die blijven worden wel gebeld (krijgen onterecht het goedkope tarief).

In de dataframe zijn de waarden per prestatiemaat te zien:
- De accuracy van het model is 0,91.
- De recall van het model is 0,70.
- De precision van het model is 0,71.
- De F1-score van het model is 0,71.

Het model voorspelt over het algemeen goed of klanten willen blijven of willen opzeggen die te zien is aan de  hoge accuracy. We zien dat het model de recall, precision en F1-score bijna gelijk zijn aan elkaar. Het probleem van dit model is dat er 91 klanten onterecht het goedkope tarief krijgen en 94 klanten die verloren gaan, omdat ze niet gebeld worden. Dit kan teveel misclassificaties zijn waardoor het model minder geschikt is om te voorspellen of klanten willen blijven of opzeggen.
